4. Как найти самое часто встречаемое значение между двумя массивами? 

````python
a = np.array([1, 2, 2, 3])
b = np.array([2, 3, 4, 2])

combined = np.concatenate((a, b))

values, counts = np.unique(combined, return_counts=True)
most_frequent_index = np.argmax(counts)

print("Самое частое значение:", values[most_frequent_index])
print("Количество повторений:", counts[most_frequent_index])
````

8. Создайте случайную матрицу 10х2 которая представляет собой декартовые координаты и преобразуйте их в полярные

````python
rng = np.random.default_rng()
coordinates = rng.random((10, 2))

x = coordinates[:, 0]
y = coordinates[:, 1]

r = np.sqrt(x**2 + y**2)
phi = np.arctan2(y, x)

polar_coordinates = np.column_stack((r, phi))

print("Декартовы координаты:")
print(coordinates)

print("Полярные координаты:")
print(polar_coordinates)
````

10. Как конвертировать тип массив от float32 в int32 in place?

````python
numbers = np.array([6.7, 5.5, -6.9], dtype=np.float32)

print("Массив до преобразования:", numbers)
print("Тип до преобразования:", numbers.dtype)

int_view = numbers.view(np.int32)

for i in range(numbers.size):
    int_view[i] = int(numbers[i])

numbers = int_view

print("Массив после преобразования:", numbers)
print("Тип после преобразования:", numbers.dtype)
````

2. Сколько процентов студентов посещают внеклассовые активности?

3. Правда ли, что в среднем девушки сдают экзамены лучше, чем мужчины

````python
average_scores = df.groupby("Gender")["Exam_Score"].mean()

female_score = average_scores["Female"]
male_score = average_scores["Male"]

print("Средний балл девушек:", female_score)
print("Средний балл юношей:", male_score)

if female_score > male_score:
    print("Ответ: да")
else:
    print("Ответ: нет")
````

4. Выведите 25 и 75 перцентили времени сна студентов

````python
sleep_25 = df["Sleep_Hours"].quantile(0.25)
sleep_75 = df["Sleep_Hours"].quantile(0.75)

print("25-й перцентиль времени сна:", sleep_25)
print("75-й перцентиль времени сна:", sleep_75)
````

5. Правда ли, что те, кто спит больше(75 перцентиль и выше) сдает экзамен лучше, чем студенты с недосыпом(25 перцентиль и меньше)

````python
more_sleep = df[df["Sleep_Hours"] >= sleep_75]
less_sleep = df[df["Sleep_Hours"] <= sleep_25]

more_sleep_score = more_sleep["Exam_Score"].mean()
less_sleep_score = less_sleep["Exam_Score"].mean()

print(f"Средний балл спящих от {sleep_75} часов:", more_sleep_score)
print(f"Средний балл спящих до {sleep_25} часов включительно:", less_sleep_score)

if more_sleep_score > less_sleep_score:
    print("Ответ: да")
else:
    print("Ответ: нет")
````
