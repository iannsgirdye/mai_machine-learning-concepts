4. Как найти самое часто встречаемое значение между двумя массивами? 

````python
a = np.array([1, 2, 2, 3])
b = np.array([2, 3, 4, 2])

combined = np.concatenate((a, b))

values, counts = np.unique(combined, return_counts=True)
most_frequent_index = np.argmax(counts)

print("Самое частое значение:", values[most_frequent_index])
print("Количество повторений:", counts[most_frequent_index])
````

8. Создайте случайную матрицу 10х2 которая представляет собой декартовые координаты и преобразуйте их в полярные

````python
rng = np.random.default_rng()
coordinates = rng.random((10, 2))

x = coordinates[:, 0]
y = coordinates[:, 1]

r = np.sqrt(x**2 + y**2)
phi = np.arctan2(y, x)

polar_coordinates = np.column_stack((r, phi))

print("Декартовы координаты:")
print(coordinates)

print("Полярные координаты:")
print(polar_coordinates)
````

10. Как конвертировать тип массив от float32 в int32 in place?

````python
numbers = np.array([6.7, 5.5, -6.9], dtype=np.float32)

print("Массив до преобразования:", numbers)
print("Тип до преобразования:", numbers.dtype)

int_view = numbers.view(np.int32)

for i in range(numbers.size):
    int_view[i] = int(numbers[i])

numbers = int_view

print("Массив после преобразования:", numbers)
print("Тип после преобразования:", numbers.dtype)
````

1. Посчитайте среднее кол-во часов уделяемое на сон студентами

````python
average_sleep = df["Sleep_Hours"].mean()

print("Среднее количество часов сна:", average_sleep)
````

2. Сколько процентов студентов посещают внеклассовые активности?

````python
students_with_activities = (df["Extracurricular_Activities"] == "Yes").sum()

percentage = students_with_activities / len(df) * 100

print("Процент студентов, посещающих внеклассные активности:", percentage, "%")
````

3. Правда ли, что в среднем девушки сдают экзамены лучше, чем мужчины

````python
average_scores = df.groupby("Gender")["Exam_Score"].mean()

female_score = average_scores["Female"]
male_score = average_scores["Male"]

print("Средний балл девушек:", female_score)
print("Средний балл юношей:", male_score)

if female_score > male_score:
    print("Ответ: да")
else:
    print("Ответ: нет")
````

4. Выведите 25 и 75 перцентили времени сна студентов

````python
sleep_25 = df["Sleep_Hours"].quantile(0.25)
sleep_75 = df["Sleep_Hours"].quantile(0.75)

print("25-й перцентиль времени сна:", sleep_25)
print("75-й перцентиль времени сна:", sleep_75)
````

5. Правда ли, что те, кто спит больше(75 перцентиль и выше) сдает экзамен лучше, чем студенты с недосыпом(25 перцентиль и меньше)

````python
more_sleep = df[df["Sleep_Hours"] >= sleep_75]
less_sleep = df[df["Sleep_Hours"] <= sleep_25]

more_sleep_score = more_sleep["Exam_Score"].mean()
less_sleep_score = less_sleep["Exam_Score"].mean()

print(f"Средний балл спящих от {sleep_75} часов:", more_sleep_score)
print(f"Средний балл спящих до {sleep_25} часов включительно:", less_sleep_score)

if more_sleep_score > less_sleep_score:
    print("Ответ: да")
else:
    print("Ответ: нет")
````

6. Влияет ли физическая активность на результаты экзамена?

````python
correlation = df["Physical_Activity"].corr(df["Exam_Score"])
correlation_threshold = 0.3

print("Корреляция:", correlation)

if abs(correlation) >= correlation_threshold:
    print("Ответ: да")
else:
    print("Ответ: нет")
````

7. Правда ли, что слуденты, чьи родители имеют высшее образование(College || Postgraduate), сдают экзамены лучше, чем остальные?

````python
education = df["Parental_Education_Level"]
has_higher_education = education.isin(["College", "Postgraduate"])

higher_score = df[has_higher_education]["Exam_Score"].mean()
other_score = df[education.notna() & ~has_higher_education]["Exam_Score"].mean()

print("Средний балл при высшем образовании родителей:", higher_score)
print("Средний балл при другом образовании родителей:", other_score)

if higher_score > other_score:
    print("Ответ: да")
else:
    print("Ответ: нет")
````

8. Как влияет участие родителей на результаты экзаменов

````python
parental_scores = df.groupby("Parental_Involvement")["Exam_Score"].mean()

print(parental_scores)

if parental_scores["Low"] < parental_scores["Medium"] < parental_scores["High"]:
    print("Ответ: с ростом участия родителей средний балл увеличивается")
elif parental_scores["Low"] > parental_scores["Medium"] > parental_scores["High"]:
    print("Ответ: с ростом участия родителей средний балл уменьшается")
else:
    print("Ответ: последовательного роста или снижения среднего балла нет")
````

9. Правда ли, что в семьях с доходом больше среднего дети сдают экзамены лучше?

````python
income_scores = df.groupby("Family_Income")["Exam_Score"].mean()

print("Средний балл по уровню дохода семьи:")
print(income_scores)

if (income_scores["High"] > income_scores["Medium"] and income_scores["High"] > income_scores["Low"]):
    print("Ответ: да")
else:
    print("Ответ: нет")
````

10. Посчитайте долю студентов сдавших экзамен на более чем 70 баллов в семьях с низким доходом и отсутствием доступа к интернету

````python
selected_students = df[(df["Family_Income"] == "Low") & (df["Internet_Access"] == "No")]

passed_count = (selected_students["Exam_Score"] > 70).sum()
share = passed_count / len(selected_students)

print("Количество студентов в группе:", len(selected_students))
print("Набрали больше 70 баллов:", passed_count)
print("Доля:", share)
````

3. Как зависит результат конечного экзамена от результата первого экзамена

4. Постройте box plot отображающий распределение количества сна детей сдавших экзамены на оценку ниже и выше 85 баллов

````python
score_threshold = 85

sleep_below = df[df["Exam_Score"] < score_threshold]["Sleep_Hours"]
sleep_above = df[df["Exam_Score"] > score_threshold]["Sleep_Hours"]

plt.boxplot([sleep_below, sleep_above], tick_labels=[f"Ниже {score_threshold} баллов", f"Выше {score_threshold} баллов"])

plt.title("Продолжительность сна по результатам экзамена")
plt.ylabel("Количество часов сна в сутки")

plt.show()
````
